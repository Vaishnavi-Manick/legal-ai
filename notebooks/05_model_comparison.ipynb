# AILA 2019 Model Comparison & Evaluation Summary

This notebook compares evaluation performance across all implemented legal relevance models on the AILA 2019 test set (Q11-Q50).

### Models Compared:
1. **BiLSTM**: Siamese-style Dual Encoder with Attention Pooling.
2. **Transformer**: Cross-Encoder with Sliding-Window Chunking and Max-Pooling Aggregation.
3. **Model 3**: Upcoming Architecture (TBD).

### Key Information Retrieval Metrics:
- **P@5**: Precision at rank 5
- **Recall@5**: Recall at rank 5
- **MRR**: Mean Reciprocal Rank
- **MAP**: Mean Average Precision
- **NDCG@5**: Normalized Discounted Cumulative Gain at rank 5
- **NDCG@10**: Normalized Discounted Cumulative Gain at rank 10

```python
# Cell 1: Environment & Imports
import os
import sys
import pandas as pd
import matplotlib.pyplot as plt

project_root = os.path.abspath(os.path.join(os.getcwd(), '..')) if os.path.basename(os.getcwd()) == 'notebooks' else os.path.abspath('.')
comparison_csv_path = os.path.join(project_root, 'results', 'comparison', 'model_comparison.csv')
```

```python
# Cell 2: Load Model Comparison Summary Table
df_comp = pd.read_csv(comparison_csv_path)
print('========================================================================')
print('                   AILA 2019 MODEL COMPARISON SUMMARY Table             ')
print('========================================================================')
df_comp
```

```python
# Cell 3: Comparative Bar Chart Visualization (BiLSTM vs Transformer)
# Filter out TBD rows for plotting
eval_df = df_comp[df_comp['P@5'] != 'TBD'].copy()
metric_cols = ['P@5', 'Recall@5', 'MRR', 'MAP', 'NDCG@5', 'NDCG@10']
for col in metric_cols:
    eval_df[col] = eval_df[col].astype(float)

ax = eval_df.set_index('Model')[metric_cols].T.plot(kind='bar', figsize=(12, 6), width=0.7)
plt.title('TREC Evaluation Metrics Comparison: BiLSTM vs Transformer', fontsize=14, fontweight='bold')
plt.ylabel('Score', fontsize=12)
plt.xlabel('Metric', fontsize=12)
plt.grid(axis='y', linestyle=':', alpha=0.7)
plt.legend(title='Model Architecture', fontsize=11)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
